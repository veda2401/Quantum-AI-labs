```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram

sim = AerSimulator()

def oracle(qc):
    qc.cz(0, 1)

def diffuser(qc):
    qc.h([0, 1])
    qc.x([0, 1])
    qc.cz(0, 1)
    qc.x([0, 1])
    qc.h([0, 1])

def grover(iterations):
    qc = QuantumCircuit(2)

    qc.h([0, 1])

    for i in range(iterations):
        oracle(qc)
        diffuser(qc)

    qc.measure_all()

    return qc

qc1 = grover(1)
result1 = sim.run(qc1, shots=1000).result()
counts1 = result1.get_counts()

qc3 = grover(3)
result3 = sim.run(qc3, shots=1000).result()
counts3 = result3.get_counts()

plot_histogram(
    [counts1, counts3],
    legend=["1 Iteration", "3 Iterations"]
)
```
